```python
import pdfplumber

dataPath = "참고_품질검수보고서__텍스트자료임.pdf"

with pdfplumber.open(dataPath) as pdf:
    for i, page in enumerate(pdf.pages):
        print(f"📄 [페이지 {i}] ------------------------------------")

        # 🔹 1. 표(table) 추출
        tables = page.extract_tables()
        for t_idx, table in enumerate(tables):
            print(f"  ▪ 표 {t_idx+1}:")
            for row in table:
                print(row)
            print()

        # 🔹 2. 텍스트 레이아웃(좌표 포함)
        words = page.extract_words()
        print(f"  ▪ 텍스트 레이아웃(샘플 5개):")
        for w in words[:5]:
            print(w)
        print("="*80)
```

```python
##########################333
## 페이지를 json 구조로 정리

import json

result = []

with pdfplumber.open(dataPath) as pdf:
    for i, page in enumerate(pdf.pages):
        page_info = {"page": i, "tables": [], "words": []}

        # 표
        for table in page.extract_tables():
            page_info["tables"].append(table)

        # 레이아웃 단어
        page_info["words"] = page.extract_words()

        result.append(page_info)

# JSON으로 저장
with open("pdf_layout.json", "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)
```

### [2-3] 하이브리드 (랭체인 로더 + 일반 => 도큐먼트 객체)

```python
from langchain_core.documents import Document
from langchain_community.document_loaders.base import BaseLoader
import pdfplumber

class PDFPlumberTableLoader(BaseLoader):
    def __init__(self, file_path: str):
        self.file_path = file_path

    def load(self):
        docs = []
        with pdfplumber.open(self.file_path) as pdf:
            total_pages = len(pdf.pages)
            for i, page in enumerate(pdf.pages):
                text = page.extract_text() or ""
                tables = page.extract_tables()
                words = page.extract_words()

                metadata = {
                    "source": self.file_path,
                    "page": i,
                    "total_pages": total_pages,
                    "num_tables": len(tables),
                }

                page_content = {
                    "text": text,
                    "tables": tables,
                    "words": words[:10],  # 미리보기용 (너무 많으면 메모리 큼)
                }

                docs.append(Document(page_content=str(page_content), metadata=metadata))
        return docs
```

```python
dataPath = "참고_품질검수보고서__텍스트자료임.pdf"

loader = PDFPlumberTableLoader(dataPath)
docs = loader.load()

for d in docs:
    print(f"📄 페이지 {d.metadata['page']} (표 {d.metadata['num_tables']}개)")
    print(d.page_content[:300], "\n")
```

## [3]최종
 ### 표추출 자료 & 도큐먼트 객체에 넣어 RAG

```python
# 필수 패키지 설치
!pip install langchain langchain-community langchain-openai pdfplumber chromadb openai --break-system-packages
```

```python
from langchain_core.documents import Document
from langchain_community.document_loaders.base import BaseLoader
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import Chroma
from langchain.retrievers import ContextualCompressionRetriever
from langchain.retrievers.document_compressors import LLMChainExtractor
from langchain.text_splitter import RecursiveCharacterTextSplitter
import pdfplumber

# ============================================
# 1. PDF 로더 (테이블 포함)
# ============================================
class PDFPlumberTableLoader(BaseLoader):
    def __init__(self, file_path: str):
        self.file_path = file_path

    def load(self):
        docs = []
        with pdfplumber.open(self.file_path) as pdf:
            for i, page in enumerate(pdf.pages):
                text = page.extract_text() or ""
                tables = page.extract_tables()

                # 테이블을 텍스트로 변환
                table_text = ""
                for table in tables:
                    for row in table:
                        table_text += " | ".join([str(cell) if cell else "" for cell in row]) + "\n"

                # 텍스트 + 테이블 합치기
                full_content = f"{text}\n\n[표 데이터]\n{table_text}" if table_text else text

                metadata = {
                    "source": self.file_path,
                    "page": i + 1,
                    "num_tables": len(tables),
                }

                docs.append(Document(page_content=full_content, metadata=metadata))
        return docs

# ============================================
# 2. 문서 로드 및 청킹
# ============================================
file_path = "참고_품질검수보고서__텍스트자료임.pdf"

# 문서 로드
loader = PDFPlumberTableLoader(file_path)
docs = loader.load()

print(f"✅ 총 {len(docs)}개 페이지 로드 완료\n")

# 청킹
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
chunks = text_splitter.split_documents(docs)

print(f"✅ 총 {len(chunks)}개 청크 생성 완료\n")

# ============================================
# 3. 벡터 DB 생성 (MMR 리트리버)
# ============================================
api_key = "YOUR_OPENAI_API_KEY"

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small",
    openai_api_key=api_key
)

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="quality_reports"
)

base_retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 5,
        "fetch_k": 20,
        "lambda_mult": 0.7
    }
)

print("✅ 벡터 DB 및 MMR 리트리버 생성 완료\n")

# ============================================
# 4. 압축 리트리버 (선택사항)
# ============================================
llm = ChatOpenAI(
    temperature=0,
    model="gpt-4o-mini",
    openai_api_key=api_key
)

compressor = LLMChainExtractor.from_llm(llm)

final_retriever = ContextualCompressionRetriever(
    base_compressor=compressor,
    base_retriever=base_retriever
)

print(" 압축 리트리버 생성 완료\n")

# ============================================
# 5. 검색 테스트
# ============================================
query = "커패시터 소손 제품의 조치사항은?"

results = final_retriever.get_relevant_documents(query)

print(f" 질문: {query}\n")
print(f" 검색 결과 ({len(results)}개):\n")

for i, doc in enumerate(results, 1):
    print(f"[결과 {i}]")
    print(f"페이지: {doc.metadata.get('page', 'N/A')}")
    print(f"내용:\n{doc.page_content}\n")
    print("-" * 80)
```

### [3-3] LLM & RAG & 표포함

```python
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate

# 프롬프트 템플릿
template = """다음 품질검수 보고서를 참고하여 질문에 답변해주세요.

문서 내용:
{context}

질문: {question}

답변 시 주의사항:
- 표의 데이터를 정확히 인용하세요
- 시리얼 번호, Lot 번호를 명시하세요
- 수치는 단위와 함께 표시하세요
- 출처 페이지를 밝히세요

답변:"""

prompt = PromptTemplate(
    template=template,
    input_variables=["context", "question"]
)

# QA 체인 생성
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=final_retriever,
    chain_type_kwargs={"prompt": prompt},
    return_source_documents=True
)

# 질문
result = qa_chain.invoke({
    "query": "최대 소음이 40dB 이하인 제품의 시리얼 번호와 검수 담당자는?"
})

print("답변:", result['result'])
print("\n출처:")
for doc in result['source_documents']:
    print(f"- 페이지 {doc.metadata['page']}")
```

```python
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate

# ============================================
# 1. 프롬프트 템플릿 설정
# ============================================
template = """당신은 품질검수 보고서 분석 전문가입니다.
아래 문서를 참고하여 질문에 정확하게 답변하세요.

문서 내용:
{context}

질문: {question}

답변 규칙:
- 표의 데이터를 정확히 참조하세요
- 시리얼 번호, Lot 번호를 명시하세요
- 수치는 단위와 함께 표시하세요
- 여러 제품이 있으면 모두 나열하세요
- 답을 모르면 "문서에서 찾을 수 없습니다"라고 하세요

답변:"""

prompt = PromptTemplate(
    template=template,
    input_variables=["context", "question"]
)

# ============================================
# 2. QA 체인 생성
# ============================================
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=final_retriever,
    chain_type_kwargs={"prompt": prompt},
    return_source_documents=True
)

# ============================================
# 3. 표 관련 질문 리스트
# ============================================
table_specific_queries = [
    # 체크리스트 기반
    "외관 검사 결과가 '양호'인 제품은?",
    "기능 검사에서 '불가' 항목이 있는 제품은?",

    # 측정값 비교
    "기준값 대비 측정값이 우수한 항목들은?",
    "기준 초과한 제품의 비고란에는 뭐라고 적혀있나?",

    # 결과 분류
    "최종 판정이 '합격'인 제품 개수는?",
    "반품 처리된 제품의 사유는?",
]

# ============================================
# 4. 질문 실행 및 결과 출력
# ============================================
for i, query in enumerate(table_specific_queries, 1):
    print(f"\n{'='*80}")
    print(f"📋 질문 {i}: {query}")
    print('='*80)

    # QA 체인 실행
    result = qa_chain.invoke({"query": query})

    # 답변 출력
    print(f"\n💡 답변:\n{result['result']}")

    # 출처 문서 출력
    print(f"\n📄 출처:")
    for doc in result['source_documents'][:2]:  # 상위 2개만
        print(f"  - 페이지 {doc.metadata['page']}")
        print(f"    내용 미리보기: {doc.page_content[:150]}...\n")
```
